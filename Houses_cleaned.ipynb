```python
df=pd.read_csv("houses.csv")
```

```python
df=df.drop_duplicates()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df=df.drop(columns=['link','property_id'])
```

```python
df.head()
```

```python
df=df.rename(columns={'rate':'price_per_sqft'})
df.head()
```

```python
df['society'].value_counts()
```

```python
df['society'].value_counts().shape
```

```python
# abb basically humara kaam hai society ke naam me se ratings jese 4.5 ★, 3★, 5 ★ 
# hataana, phir text ko clean & lowercase karna.
import re
df['society'] = df['society'].apply(lambda name: re.sub(r'\d+(\.\d+)?\s?★', '', str(name)).strip()).str.lower()
```

```python
df['society']=df['society'].str.replace('nan','independent')
```

```python
df.shape
```

```python
df.info()
```

```python
df['price'].value_counts()
```

```python
# now some of the rows in the price were having the value as price on request

df = df[df['price'] != 'Price on Request']
```

```python
# now converting the price to crore as some of them were in lac
def treat_price(x):
    if type(x)==float:
        return x
    else:
        if x[1]=='Lac':
            return round(float(x[0])/100,2)
        else:
            return round(float(x[0]),2)
```

```python
df['price'] = df['price'].str.split(' ').apply(treat_price)
```

```python
#  abb hum log rice_per_sqft par operations karenge 
df['price_per_sqft'].value_counts()
```

```python
# here we are cleaning the price_per_sqft column 
df['price_per_sqft']=df['price_per_sqft'].str.split('/').str.get(0).str.replace('₹','').str.replace(',','').str.strip().astype('float')
```

```python
df.head()
```

```python
df['bedRoom'].value_counts()
```

```python
df[df['bedRoom'].isnull()]
```

```python
df=df[~df['bedRoom'].isnull()]
df.shape

df['bedRoom'] = df['bedRoom'].str.split(' ').str.get(0).astype('int')
```

```python
df.head()
```

```python
df['bathroom'].value_counts()
```

```python
df['bathroom'].isnull().sum()
```

```python
df['bathroom']=df['bathroom'].str.split(' ').str.get(0).astype('int')
```

```python
df['balcony'].value_counts()
df['balcony']=df['balcony'].str.split(' ').str.get(0).str.replace('No','0')
```

```python
df.head()
```

```python
df['additionalRoom'].value_counts()
df['additionalRoom'].isnull().sum()
```

```python
df['additionalRoom']=df['additionalRoom'].str.lower()
df.head()
```

```python
df['noOfFloor'].value_counts()
df['noOfFloor']=df['noOfFloor'].str.split(' ').str.get(0)
df.head()
```

```python
# changing the name from noOfFloor to floorNum
df.rename(columns={'noOfFloor':'floorNum'},inplace=True)
```

```python
df.head()
```

```python
# filling the missing values in the facing section 
df['facing'].fillna('NA',inplace=True)
```

```python
df['area']=round((df['price']*10000000)/df['price_per_sqft'])
```

```python
df.insert(loc=1,column='property_type',value='house')
```
